**An quick example of to remember quiver plots**

```python
import numpy as np
import matplotlib.pyplot as plt

# Read the arrays vertically – there are four arrows
# (X, Y) are starting points of the arrows
# (U, V) are the direction / offset points
# A Quiver plot draws arrows from (X, Y) to (X+U, Y+V)
X = np.array([1, 1, 1, 1])
Y = np.array([1, 1, 1, 1])
U = np.array([-0.5, 0.5, -0.5, 0.5])  
V = np.array([1, 1, -1, -1])   

plt.quiver(X, Y, U, V, angles='xy', scale_units='xy', scale=1)
plt.xlim(-2, 2)
plt.ylim(-2, 2)
plt.grid(True)
plt.show()
```

**A simple example vector field of a swirl**

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-2, 2, 10)
y = np.linspace(-2, 2, 10)
X, Y = np.meshgrid(x, y)   # build a grid of points

U = -Y   # vector field that swirls around the origin
V = X

plt.quiver(X, Y, U, V)
plt.gca().set_aspect('equal')
plt.show()
```

**Example of a velocity field in a channel**

```python
import numpy as np
import matplotlib.pyplot as plt

# Grid of points across the channel
x = np.arange(0, 8, 2)
y = np.arange(-1, 1, 0.2)
X, Y = np.meshgrid(x, y)

# The velocity field
U = 1 - Y**2     # rightward speed: max at center (y=0), zero at walls (y=±1)
V = np.zeros_like(X)   # no vertical motion anywhere
# V = 0.2 * np.ones_like(X)

plt.quiver(X, Y, U, V, angles='xy', scale_units='xy', scale=1)
plt.axhline(1, color='gray')    # top wall
plt.axhline(-1, color='gray')   # bottom wall
plt.gca().set_aspect('equal')
plt.show()
```
